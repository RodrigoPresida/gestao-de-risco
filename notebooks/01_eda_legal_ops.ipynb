# Análise Exploratória: Legal Ops Strategic Analytics

Este notebook apresenta a análise inicial dos dados enriquecidos com métricas de Legal Ops.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Carregar dados
df = pd.read_csv('../data/legal_ops_dataset.csv')
df.head()
```

## 1. Visão Geral da Carteira

```python
fig = px.pie(df, names='risco_cpc25', title='Distribuição de Risco (CPC 25)', hole=0.4)
fig.show()
```

## 2. Exposição Financeira (Exposure) vs Provisão

```python
financial_summary = df.groupby('risco_cpc25').agg({
    'valor_causa': 'sum',
    'valor_provisao': 'sum'
}).reset_index()

fig = go.Figure(data=[
    go.Bar(name='Valor da Causa (Exposure)', x=financial_summary['risco_cpc25'], y=financial_summary['valor_causa']),
    go.Bar(name='Valor Provisionado', x=financial_summary['risco_cpc25'], y=financial_summary['valor_provisao'])
])
fig.update_layout(barmode='group', title='Exposure vs Provisão por Classe de Risco')
fig.show()
```
